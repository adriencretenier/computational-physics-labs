```python
"""Importation des modules"""
import matplotlib.pyplot as plt
import numpy as np
```

```python
#Question 1 :

P = [0]
Xp = [0.1]
for i in range(1,300):
    Xp.append(2*(1-Xp[i-1])*Xp[i-1])
    P.append(i)

plt.plot(P,Xp)
plt.xlabel('Génération P')
plt.ylabel('Population')
plt.title("Evolution de la population d'insecte en fonction de la génération")
plt.show()
```

```python
#Question 2 :

def Evolution(r,pmax,x0):
    P = [0]
    Xp = [x0]
    for i in range(1,pmax+1):
        xi = r*(1-Xp[i-1])*Xp[i-1]
        Xp.append(xi)
        P.append(i)
    return[P,Xp]

plt.figure(figsize=(17,14))
plt.subplot(3,3,1)
X1 = Evolution(1,100,0.1)[0]
Y1 = Evolution(1,100,0.1)[1]
plt.plot(X1,Y1)
plt.xlabel('Génération P')
plt.ylabel('Population')
plt.title("Evolution pour r = 1")

plt.subplot(3,3,2)
X2 = Evolution(2,100,0.1)[0]
Y2 = Evolution(2,100,0.1)[1]
plt.plot(X2,Y2)
plt.xlabel('Génération P')
plt.ylabel('Population')
plt.title("Evolution pour r = 2")

plt.subplot(3,3,3)
X3 = Evolution(2.7,100,0.1)[0]
Y3 = Evolution(2.7,100,0.1)[1]
plt.plot(X3,Y3)
plt.xlabel('Génération P')
plt.ylabel('Population')
plt.title("Evolution pour r = 2,7")

plt.subplot(3,3,4)
X4 = Evolution(3,100,0.1)[0]
Y4 = Evolution(3,100,0.1)[1]
plt.plot(X4,Y4)
plt.xlabel('Génération P')
plt.ylabel('Population')
plt.title("Evolution pour r = 3")

plt.subplot(3,3,5)
X5 = Evolution(3.3,100,0.1)[0]
Y5 = Evolution(3.3,100,0.1)[1]
plt.plot(X5,Y5)
plt.xlabel('Génération P')
plt.ylabel('Population')
plt.title("Evolution pour r = 3.3")

plt.subplot(3,3,6)
X6 = Evolution(3.6,100,0.1)[0]
Y6 = Evolution(3.6,100,0.1)[1]
plt.plot(X6,Y6)
plt.xlabel('Génération P')
plt.ylabel('Population')
plt.title("Evolution pour r = 3.6")

plt.subplot(3,3,7)
X7 = Evolution(3.7,100,0.1)[0]
Y7 = Evolution(3.7,100,0.1)[1]
plt.plot(X7,Y7)
plt.xlabel('Génération P')
plt.ylabel('Population')
plt.title("Evolution pour r = 3,7")

plt.subplot(3,3,8)
X8 = Evolution(3.9,100,0.1)[0]
Y8 = Evolution(3.9,100,0.1)[1]
plt.plot(X8,Y8)
plt.xlabel('Génération P')
plt.ylabel('Population')
plt.title("Evolution pour r = 3,9")

plt.subplot(3,3,9)
X9 = Evolution(4,100,0.1)[0]
Y9 = Evolution(4,100,0.1)[1]
plt.plot(X9,Y9)
plt.xlabel('Génération P')
plt.ylabel('Population')
plt.title("Evolution pour r = 4")

plt.show()
```

```python
#Question 3 :

def Visualisation(r,pmax,x0):
        
    """Représentation de F"""
    X = np.linspace(0,1,100)
    F = []    
    for i in X:
        F.append(r*(1-i)*i)
        
    """Représentation de l'escalier"""
    Xe = [x0,x0]
    E = [0,r*(1-x0)*x0]
    for i in range(2,len(F)-1):
        Xe.append(r*(1-Xe[len(Xe)-1])*Xe[len(Xe)-1])
        Xe.append(r*(1-Xe[len(Xe)-2])*Xe[len(Xe)-2])
        E.append(E[len(E)-1])
        E.append(r*(1-E[len(E)-1])*E[len(E)-1])
    plt.plot(X,F,color='blue')
    plt.plot(Xe,E,color='red')
    plt.plot(X,X,color='green')
    
    
plt.figure(figsize=(17,14))
plt.subplot(2,2,1)
print(Visualisation(2,100,0.1))
plt.title("Evolution pour r = 2")

plt.subplot(2,2,2)
print(Visualisation(3.3,100,0.1))
plt.title("Evolution pour r = 3,3")

plt.subplot(2,2,3)
print(Visualisation(3.6,100,0.1))
plt.title("Evolution pour r = 3,6")

plt.subplot(2,2,4)
print(Visualisation(3.9,100,0.1))
plt.title("Evolution pour r = 3,9")

plt.show()
    
```

```python
#Question 4 :

R1 = np.linspace(2,4,200)
R = []
Y = []
for r in R1:
    Xp = Evolution(r,2000,0.1)[1]
    if r<=3 :
        for i in range(1,101):
            Y.append(Xp[len(Xp)-i])
            R.append(r)
    if r>3 :
        for i in range(1,1001):
            Y.append(Xp[len(Xp)-i])
            R.append(r)
            
plt.figure(figsize=(10,10))
plt.plot(R,Y,'+',markersize=1)
plt.xlabel('r')
plt.ylabel('Population')
plt.title("Valeur des populations en fonction de r")

plt.show()
```
